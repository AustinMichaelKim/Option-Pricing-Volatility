# Black–Scholes–Merton Model

이 노트북은 BSM PDE의 closed-form solution으로 유럽형 call/put 가격을 계산한다. 기초자산은 위험중립 측도에서 GBM을 따르고, 만기까지 연속복리 무위험이자율 $r$, 연속배당수익률 $q$, 변동성 $\sigma$가 일정하며 거래마찰이 없다고 가정한다. 시간 $T$는 연 단위이고 $r$, $q$, $\sigma$는 소수로 입력한다.

## Closed-form solution

만기가 양수이고 변동성이 양수일 때

$$
d_1 = \frac{\ln(S/K) + (r-q+\tfrac{1}{2}\sigma^2)T}{\sigma\sqrt{T}},
\qquad
d_2 = d_1 - \sigma\sqrt{T}
$$

이며, 표준정규 누적분포함수를 $N(\cdot)$라 하면 가격은

$$
C = Se^{-qT}N(d_1) - Ke^{-rT}N(d_2),
\qquad
P = Ke^{-rT}N(-d_2) - Se^{-qT}N(-d_1)
$$

이다. 실제 계산은 공식을 노트북에 다시 구현하지 않고 `src` 패키지의 `bsm_price`를 사용한다.

```python
# package 함수 import
import pandas as pd

from option_pricing_volatility.models.bsm import bsm_price
```

```python
# 기존 Binomial 및 GBM-MC 노트북과 동일한 synthetic call/put 예제
synthetic_parameters = {
    "spot": 100.0,
    "strike": 100.0,
    "maturity": 1.0,
    "rate": 0.05,
    "volatility": 0.2,
    "dividend_yield": 0.0,
}

bsm_results = pd.DataFrame(
    [
        {
            "option_type": option_type,
            "bsm_price": bsm_price(
                **synthetic_parameters, option_type=option_type
            ),
        }
        for option_type in ("call", "put")
    ],
    columns=["option_type", "bsm_price"],
)

# 기존 수렴 결과와 option_type 열을 기준으로 병합할 수 있는 tidy 결과
bsm_results
```
